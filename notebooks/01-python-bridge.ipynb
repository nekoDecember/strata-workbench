# Strata Python Bridge
画面でデータを読み込み、選択してから実行してください。トークンは環境から取得します。

```python
from strata.sdk import Client
client = Client.from_env()
datasets = client.datasets()
[(d['id'], d['name']) for d in datasets]
```

```python
dataset_id = datasets[0]['id']
df = client.frame(dataset_id)
selected = client.frame(dataset_id, selected=True)
print(f'対象 {len(df):,} 行 / 選択 {len(selected):,} 行')
display(df.head())
```

```python
# 統計や加工は、通常のPythonとして書けます。
display(df.select_dtypes('number').drop(columns='__row_id', errors='ignore').describe())
```
